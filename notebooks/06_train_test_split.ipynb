# Split the data into training and test sets

### Split the data 70/30 for training and test sets, respectively. 

We will do model calibration and fitting on the training set then save the test set at the end to compare each of the calibrated models.

```python

# Chronologically split the combined NBA dataset into train/test CSVs.


from pathlib import Path
import pandas as pd


# ── configuration ────────────────────────────────────────────────
RAW_FILE   = Path("../data/processed/no_missing_538_kaggle_nba.csv")
OUT_DIR    = Path("../data/processed")
TRAIN_FILE = OUT_DIR / "train_set.csv"
TEST_FILE  = OUT_DIR / "test_set.csv"
SPLIT_FRAC = 0.70            # 70 % → train, 30 % → test
# ─────────────────────────────────────────────────────────────────


def main() -> None:
    # 1 . load & clean
    df = (
        pd.read_csv(RAW_FILE, parse_dates=["date_game"])  # parse_dates does the conversion for you
          .sort_values("date_game")                       # guarantee chronological order
          .set_index("date_game")                         # date as index (handy for later)
    )

    # 2 . chronological split
    split_idx = int(len(df) * SPLIT_FRAC)
    train_df  = df.iloc[:split_idx].reset_index()        # reset → keep the game_date column
    test_df   = df.iloc[split_idx:].reset_index()

    # 3 . persist
    OUT_DIR.mkdir(parents=True, exist_ok=True)            # create folder if missing
    train_df.to_csv(TRAIN_FILE, index=False)
    test_df.to_csv(TEST_FILE,  index=False)

    # 4 . quick sanity printout
    print(f"✓ Train rows : {len(train_df):,}  →  {TRAIN_FILE}")
    print(f"✓ Test  rows : {len(test_df):,}  →  {TEST_FILE}")


if __name__ == "__main__":
    main()
```
